## Question 1: Choose a case

Title: Incident 37: Female Applicants Down-Ranked by Amazon Recruiting Tool,
URL: https://incidentdatabase.ai/cite/37/

In 2014 five of Amazon's machine learning employees developed an AI system that helps automatically sorting CV's of the candidates. They wanted to reach the top candidates with that algorithm. However, there was a big issue with the algorithm. The system was trained on data over 10 year but this data was including mostly the men data. As a result, their system was prefering male candidates more and it was downgrading the women's resumes when their CV was including the word "women" or "women's". They tried to fix the algorithm but at the end in 2017 they stopped working on the algorithm. Once this issue was identified, Amazon decided to discontinue this system, highlighting how difficult it is to ensure the fairness equality in the AI based hiring process.

The candidates who were applying to Amazon were affected by this incident. The female applicants were the primarily group affected by this incident for applying to the technical roles at Amazon. As an example if "women who code" was in the CV, the applicant would be penalized for it. It also was downgrading resumes if there were women-only colleges. Moreover, the algorithm was giving more scores if the CV was including some verbs that are more categorized as "macho", such as "executed", "captured" etc.

To provide fairness in such algorithms are challenging according to the experience of Amazon on their AI tool. To improve on this issue, Amazon could have incorporated the biasing audits and constraints during the model training. Regular health checks against possible bias can be done with human moderation.
